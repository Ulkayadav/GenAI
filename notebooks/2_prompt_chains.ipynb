```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
prompt=[
    {"role":"system","content":"You are a maths teacher"},
    {"role":"user","content":"What is your name?"}
]
res=llm.invoke(prompt)
res.content
```

## Prompt template - Dynamic prompts

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite") 
out=StrOutputParser()
def transform_case(result):
    return result.upper()
prompts=ChatPromptTemplate.from_messages([
    {"role":"system","content":"You are a {subject} teacher"},
    {"role":"user","content":"{question}"}
])
```

```python
final_prompt=prompts.format_messages(subject="maths",question="What is your name?")
res=llm.invoke(final_prompt)
res.content
```
