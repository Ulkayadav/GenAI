## Structured output

```python
from pydantic import BaseModel,Field
from typing import List
class ResponseStructure(BaseModel):
    name: str=Field(description="Name of the User")
    profession: str=Field(description="Profession of the User")
    age: int=Field(description="Age of the User")
    email: str=Field(description="Email address of the User")

structured_llm=llm.with_structured_output(ResponseStructure)    
res=structured_llm.invoke(f"Please give me name , profession, age and email of the text,{text}")
print(res)
print(res.name)
print(res.profession)
print(res.age)
print(type(res.email))
print(type(res))
```

```python
class Movies(BaseModel):
    title:str=Field(description="Title of the Movie")
    year:int=Field(description="Release year of the Movie")

movie_llm=llm.with_structured_output(Movies)
res=movie_llm.invoke("Tell me three bestest romantic movie")
print(res)
```

## Multiple movies 

```python
class AllMovies(BaseModel):
    movies:List[Movies]

movie_llm=llm.with_structured_output(AllMovies)
res=movie_llm.invoke("Tell me three bestest romantic movie")
print(res)
res
```
